```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

column_names = [
    'poisonous', 'cap_shape', 'cap_surface', 'cap_color', 'bruises',
    'odor', 'gill_attachment', 'gill_spacing', 'gill_size', 'gill_color',
    'stalk_shape', 'stalk_root', 'stalk_surface_above_ring',
    'stalk_surface_below_ring', 'stalk_color_above_ring',
    'stalk_color_below_ring', 'veil_type', 'veil_color', 'ring_number',
    'ring_type', 'spore_print_color', 'population', 'habitat'
]

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/mushroom/agaricus-lepiota.data'

full_df = pd.read_csv(url, header=None, names=column_names)
full_df.head()
```

```python
df = full_df[['poisonous', 'odor', 'cap_color']].copy()
df.columns = ['Class', 'Odor', 'Cap_Color']
df.head()
```

```python
odor_dummies = pd.get_dummies(df['Odor'], prefix='Odor')
cap_dummies = pd.get_dummies(df['Cap_Color'], prefix='Cap_Color')

print(odor_dummies.head())
print(cap_dummies.head())
```

pd.get_dummies() takes a column and turns each unique value into its own yes/no column

```python
y = df['Class']
print(y.value_counts())
```

y = how much mushroom is edible (0) or poisonous (1)

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

X_odor = odor_dummies
X_train, X_test, y_train, y_test = train_test_split(X_odor, y, test_size=0.25, random_state=42)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred_odor = knn.predict(X_test)
odor_accuracy = metrics.accuracy_score(y_test, y_pred_odor)
print(f"Odor Accuracy: {odor_accuracy:.4f}")
```

to predict poison mushroom, split the data into testing and training. use odor to predict poisonious. create classifier model. find the 5 most similar mushrooms it already knows about, and takes a majority vote. Show the model the training data + answers so it can learn patterns. Give it new data (no answers), let it guess, then check how many it got right (accuracy). we got 98.52% accuracy.

```python
print(f"Odor Accuracy:      {odor_accuracy:.4f}  ({odor_accuracy*100:.2f}%)")
print(f"Cap Color Accuracy: {cap_accuracy:.4f}  ({cap_accuracy*100:.2f}%)")

if odor_accuracy > cap_accuracy:
    print("\nConclusion: ODOR is the better predictor of whether a mushroom is poisonous.")
else:
    print("\nConclusion: CAP COLOR is the better predictor of whether a mushroom is poisonous.")
```
